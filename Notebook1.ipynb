```

import os
import subprocess

REPO_URL = "https://github.com/CodeCraftress/freshcart-pipelines.git"
PROJECT_DIR = "/content/freshcart-pipelines"

if not os.path.exists(PROJECT_DIR):
    subprocess.run(
        ["git", "clone", REPO_URL, PROJECT_DIR],
        check=True
    )
else:
    print("Repository already exists in Colab.")

os.chdir(PROJECT_DIR)

print("Connected repository:", os.getcwd())
print("Repository contents:")
print(os.listdir("."))
```
